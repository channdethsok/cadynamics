## Executive Summary & Core Research Objectives

Traditional 3D deep learning approaches treat Computer-Aided Design (CAD) models as **static artifacts** (unstructured point clouds, polygon meshes, or flat text sequences). In industrial engineering, however, CAD models are constructed through an **ordered sequence of parametric modeling operations** (sketches, extrusions, cuts, fillets, chamfers, and booleans).

**CADynamics** models this process as a **causal transition dynamical system**:

$$\tau = (S_0, A_0, S_1, A_1, \ldots, A_{K-1}, S_K)$$

$$\mathbf{z}_{t+1} \approx \operatorname{Predictor}(\mathbf{z}_t, A_t) \quad \text{or multi-step} \quad \mathbf{z}_{t+H} \approx \operatorname{Predictor}(\mathbf{z}_t, [A_t, \ldots, A_{t+H-1}])$$

### Why This Notebook Matters
Standard CAD datasets (like DeepCAD or ABC) only provide the final solid geometry (STEP files) or text scripts. The intermediate 3D geometric states $S_k$ are completely missing. To train a predictive world model, we must **intercept the live OpenCASCADE geometry at every single step**, extract analytical B-Rep cell complexes, ground the action parameters, and render canonical multi-view visual projections.

This notebook walks through the entire pipeline step-by-step:
1. **Raw Zero-to-CAD Parquet Inspection**: Why final STEP files are insufficient and how raw Python code is structured.
2. **AST & Runtime Interception Engine (`src.cad.tracer`)**: Dynamic method hooking and zero-disk solid interception.
3. **B-Rep Cell Complex Extraction (`src.cad.brep_extractor`)**: Face 32-D features, Edge 16-D features, UV grids, and UV-Net vs. BrepNet graph reconciliation.
4. **Action Grounding & Continuous Scaling (`src.cad.action_extractor`)**: SymLog parameterization, 2D sketch wires, and the entity selection grounding problem.
5. **Canonical Multi-View Visual Projections (`src.cad.renderer`)**: 4-view offscreen rendering with camera framing normalization.
6. **1-to-1 Sharding & Bidirectional Lineage (`src.data.schema`)**: Mathematical contracts and $O(1)$ reverse lookup.
7. **PyTorch DataLoader & Disjoint Union Collation (`src.data.dataset`)**: Variable-sized graph batching for neural training.

* Zero-to-CAD (100K human-designed mechanical parts) solves this by storing the **generative Python CadQuery script** for every part.

## Chapter 2: The Interception Engine — AST & Runtime Tracing

### The Causal Invariant
To formulate CAD modeling as a Markov Decision Process (MDP) or world model transition:
$$S_{k+1} = \mathcal{T}(S_k, A_k)$$
we must enforce a strict causal invariant across the sequence:
$$\operatorname{len}(\text{states}) = \operatorname{len}(\text{actions}) + 1$$

* $S_0$: The **empty workspace** (neutral gray canvas, zero faces, zero edges).
* $A_k$: The action context captured **immediately before** method execution (command opcode, parameters, and selected entities in $S_k$).
* $S_{k+1}$: The new solid state extracted **immediately after** the operation successfully completes in memory.

### How `CadQueryRuntimeTracer` Works
Rather than rewriting Python source code or modifying CadQuery's C++ bindings, `CadQueryRuntimeTracer` operates as a **dynamic Python context manager**:
1. It monkey-patches solid-committing operations on `cadquery.Workplane` (`extrude`, `cut`, `hole`, `fillet`, `chamfer`, etc.).
2. When an operation is called, a **pre-hook** captures $A_k$ from the current active workplane and previous shape.
3. The native CadQuery / OpenCASCADE kernel executes.
4. A **post-hook** intercepts the resulting `TopoDS_Shape` in RAM without touching the disk.

```python
# Execute the script inside CadQueryRuntimeTracer
renderer = HeadlessCadRenderer(image_size=224)

with CadQueryRuntimeTracer(render_images=True, renderer=renderer, retain_solids=True) as tracer:
    exec_globals = {'cq': cq}
    exec(raw_code, exec_globals)

states, actions = tracer.get_trajectory_data()
solids = tracer.get_solids()

print(f'Successfully traced part: {part_uuid}')
print(f'  Extracted States:  {len(states)} (S_0 ... S_{len(states)-1})')
print(f'  Extracted Actions: {len(actions)} (A_0 ... A_{len(actions)-1})')
print(f'  In-Memory Solids:  {len(solids)}')
assert len(states) == len(actions) + 1, 'Causal invariant violated: len(states) != len(actions) + 1!'
print('  Causal Invariant Verified: len(states) == len(actions) + 1')
```

```python
# Print chronological transition timeline
print('Construction History Timeline:')
print(f'  [S_0] Empty Workspace -> Faces: {states[0]["num_faces"]}, Edges: {states[0]["num_edges"]}')

for k, act in enumerate(actions):
    cmd_name = ID2CMD[act['cmd_id']]
    ref_name = REF_KIND_TO_NAME[act['ref_kind']]
    s_next = states[k + 1]
    print(f'  -- A_{k}: {cmd_name:12s} (Ref: {ref_name:14s}) --> [S_{k+1}] Faces: {s_next["num_faces"]:3d}, Edges: {s_next["num_edges"]:3d}, Vol: {s_next["volume"]:10.2f} mm^3')
```

## Chapter 3: The 3D Geometry Domain — B-Rep Cell Complex

### B-Rep Mathematical Formulation
A Boundary Representation defines a 3D solid as a hierarchical topological complex:
* **Faces (2D)**: Bounded regions on parameterized surfaces $\mathbf{S}(u, v) \in \mathbb{R}^3$.
* **Edges (1D)**: Parameterized trimming curves $\mathbf{C}(u) \in \mathbb{R}^3$ shared between incident faces.
* **Vertices (0D)**: Exact endpoints where curves meet in 3D space.

### Reconciling UV-Net and BrepNet Paradigms
In our Phase 1 literature audit ([`docs/research_logs/2026-10-01_22h.md`](../docs/research_logs/2026-10-01_22h.md)), we unified two competing paradigms in CAD geometric deep learning:

1. **UV-Net Paradigm (Face-Adjacency Graph with Aligned Directed Edges)**:
   * **Graph Nodes**: Faces ($N_{\text{faces}}$).
   * **Directed Graph Edges**: $E_{\text{adj}} = 2 \times N_{\text{edges}}$ (every boundary edge traversed forward and backward).
   * **Our Contribution**: We preserve the physical edge index in `faces_adjacency_edge_indices: LongTensor[E_adj]` and output aligned `edges_features_directed: FloatTensor[E_adj, 16]`. Furthermore, `reverse_edge_indices: LongTensor[E_adj]` enables $O(1)$ tensor-level edge fusion without dictionary lookups:
     $$\mathbf{e}_{uv} + \mathbf{e}_{vu} = \text{efeats} + \text{efeats}[\text{reverse\_edge\_indices}]$$

2. **BrepNet Paradigm (Bipartite Incidence Complex)**:
   * `face_edge_index: LongTensor[2, E_inc]` connects Face nodes to Edge nodes directly for heterogeneous alternating message passing.

Let us inspect the extracted tensors for state $S_1$ (the first generated 3D solid).

```python
# Inspect B-Rep topological tensors on State S_1
s_1 = states[1]

print('Topological Tensor Breakdown for S_1:')
print(f'  faces_features:               {s_1["faces_features"].shape}  (32 analytical features per face)')
print(f'  faces_uv:                     {s_1["faces_uv"].shape}  (7x16x16 [x,y,z,nx,ny,nz,mask] per face)')
print(f'  edges_features:               {s_1["edges_features"].shape}  (16 analytical features per physical edge)')
print(f'  edges_u:                      {s_1["edges_u"].shape}  (6x16 [x,y,z,tx,ty,tz] curve samples)')
print(f'  faces_adjacency_index:        {s_1["faces_adjacency_index"].shape}  (Directed FAG graph edges [2, E_adj])')
print(f'  edges_features_directed:      {s_1["edges_features_directed"].shape}  (Aligned edge attributes for PyG)')
print(f'  reverse_edge_indices:         {s_1["reverse_edge_indices"].shape}  (O(1) edge fusion pointers)')
print(f'  face_edge_index (bipartite):  {s_1["face_edge_index"].shape}  (Incidence matrix [2, E_inc])')

# Validate state invariants
validate_state(s_1, is_empty=False)
print('\nState S_1 successfully passed all strict mathematical invariants!')
```

```python
# Visualize the Face-Adjacency Graph (FAG)
import networkx as nx

adj_index = s_1['faces_adjacency_index'].numpy()
G = nx.DiGraph()
for f_idx in range(s_1['num_faces']):
    st_onehot = s_1['faces_features'][f_idx, :8].numpy()
    is_plane = bool(st_onehot[0] == 1.0)
    G.add_node(f_idx, is_plane=is_plane)

for edge_i in range(adj_index.shape[1]):
    u, v = int(adj_index[0, edge_i]), int(adj_index[1, edge_i])
    G.add_edge(u, v)

plt.figure(figsize=(7, 6))
pos = nx.spring_layout(G, seed=42)
node_colors = ['#38BDF8' if G.nodes[n]['is_plane'] else '#F43F5E' for n in G.nodes]

nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=600, edgecolors='#0F172A', linewidths=1.5)
nx.draw_networkx_edges(G, pos, edge_color='#64748B', arrows=True, arrowsize=14, connectionstyle='arc3,rad=0.1')
nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold', font_color='#FFFFFF')

plt.title(f'Face-Adjacency Graph (FAG) for State S_1\n({s_1["num_faces"]} Face Nodes, {adj_index.shape[1]} Directed Edges)', fontsize=12, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()
```

## Chapter 4: The Action Domain — Semantic Grounding & Continuous Scaling

### The Entity Selection Grounding Problem
When a CAD operation like `fillet(radius=2.0)` or `chamfer(length=1.5)` is invoked, standard CAD deep learning formulations only record the command and radius. But **which** edges of $S_t$ were filleted? 

Without entity selection grounding, a world model cannot know where material is being modified. We solve this by matching OpenCASCADE shapes across states using exact topological equality (`IsSame()`), producing aligned boolean masks:
* `selected_faces_mask`: `BoolTensor[Total_faces]`
* `selected_edges_mask`: `BoolTensor[Total_edges]`

### SymLog Continuous Parameter Scaling
CAD continuous parameters span extreme dynamic ranges: a chamfer might be $0.2\,\text{mm}$, while a base extrusion is $800\,\text{mm}$. Linear normalization either squashes small values or explodes large ones. We employ **Symmetric Logarithm (SymLog)**:

$$\operatorname{symlog}(x) = \operatorname{sign}(x) \cdot \ln(1 + |x|)$$

which is smooth, invertible, symmetric around zero, and preserves precision across four orders of magnitude.

```python
# Inspect an action transition step
step_idx = 0
act = actions[step_idx]

print(f'Action A_{step_idx} Detailed Breakdown:')
print(f'  Command:               {ID2CMD[act["cmd_id"]]} (ID: {act["cmd_id"]})')
print(f'  Reference Target:      {REF_KIND_TO_NAME[act["ref_kind"]]}')
print(f'  Raw Metric Params:     {act["params_raw"].numpy().round(3)}')
print(f'  SymLog Scaled Params:  {act["params"].numpy().round(3)}')
print(f'  Valid Params Mask:     {act["param_mask"].numpy().tolist()}')
print(f'  Selected Faces Count:  {int(act["selected_faces_mask"].sum())} / {len(act["selected_faces_mask"])}')
print(f'  Selected Edges Count:  {int(act["selected_edges_mask"].sum())} / {len(act["selected_edges_mask"])}')

if act.get('profile') is not None:
    prof = act['profile']
    print(f'  2D Sketch Profile:     {prof["num_wires"]} wire(s) ({prof["num_outer_loops"]} outer, {prof["num_inner_loops"]} inner)')
    print(f'  Sampled 2D Points:     {prof["points_2d"].shape}')
    print(f'  Analytic Primitives:   {len(prof.get("primitives", []))} primitive tokens (Line, Arc, Circle)')

# Validate action
validate_action(act)
print('\nAction successfully validated against schema constraints!')
```

## Chapter 5: The Visual Domain — Canonical 4-View Projections

### Why 4-View Canonical Orthographic + Isometric Projections?
* **Isometric (`iso`)**: Provides 3D spatial overview and depth perception.
* **Orthographic (`front`, `top`, `right`)**: Align directly with standard principal mechanical planes ($XZ$, $XY$, $YZ$). Unlike perspective cameras, orthographic projections preserve parallel lines, metric proportions, and eliminate perspective foreshortening.

### Camera Framing Normalization
If the camera dynamically zooms onto each solid's local bounding box at every step, adding a tiny chamfer causes the camera to stay put, but adding an extrusion causes an abrupt zoom-out jump. We eliminate scale jumps by **locking camera bounds to the global construction bounding volume**.

```python
# Render and display the visual evolution of the part across all steps
num_states_to_plot = min(len(states), 5)
fig, axes = plt.subplots(num_states_to_plot, 4, figsize=(12, 3 * num_states_to_plot))

if num_states_to_plot == 1:
    axes = np.expand_dims(axes, 0)

for s_idx in range(num_states_to_plot):
    s = states[s_idx]
    imgs = s['images'].permute(0, 2, 3, 1).numpy()  # [4, 224, 224, 3]
    op_label = 'Initial (S_0)' if s_idx == 0 else f'S_{s_idx}: {ID2CMD[actions[s_idx-1]["cmd_id"]]}'
    
    for v_idx, v_name in enumerate(CANONICAL_VIEW_NAMES):
        ax = axes[s_idx, v_idx]
        ax.imshow(imgs[v_idx])
        ax.set_xticks([])
        ax.set_yticks([])
        if s_idx == 0:
            ax.set_title(f'View: {v_name.upper()}', fontsize=11, fontweight='bold')
        if v_idx == 0:
            ax.set_ylabel(op_label, fontsize=10, fontweight='bold', labelpad=10)

plt.suptitle(f'Synchronized 4-View Visual Transition History ({part_uuid[:8]})', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

## Chapter 6: Serialization & Lineage — The 1-to-1 Shard Architecture

### Bidirectional $O(1)$ Lineage Mapping
In our data pipeline, every output `.pt` shard directly mirrors its source `.parquet` shard:
$$\text{shard\_00000.parquet} \quad \Longleftrightarrow \quad \text{shard\_00000.pt}$$

Every trajectory record stores:
* `part_id`: Unique UUID of the model.
* `source_shard`: Exact Parquet file of origin.
* `source_row`: Exact row index in the Parquet table.

This guarantees that if an anomaly or loss spike occurs during neural network training, researchers have an immediate $O(1)$ reverse lookup back to the original Python code and STEP files.

```python
# Package the trajectory container
trajectory_record = {
    'part_id': part_uuid,
    'source_split': 'val',
    'source_shard': raw_parquet_path.name,
    'source_row': 0,
    'num_steps': len(actions),
    'states': states,
    'actions': actions,
}

# Strict schema verification
validate_trajectory(trajectory_record)
print('Trajectory Container Status:')
print(f'  Part UUID:     {trajectory_record["part_id"]}')
print(f'  Source Shard:  {trajectory_record["source_shard"]}')
print(f'  Source Row:    {trajectory_record["source_row"]}')
print(f'  Num Steps (K): {trajectory_record["num_steps"]}')
print('  Trajectory passed validate_trajectory() with 100% mathematical integrity!')
```

## Chapter 7: Machine Learning Ingestion — PyTorch DataLoader & Disjoint Union

### Variable-Sized Graph Collation
Unlike images or fixed-dimension vectors, B-Rep models have variable numbers of faces and edges:
* Solid A might have 6 faces and 12 edges (a cube).
* Solid B might have 42 faces and 88 edges (a complex flange with holes and fillets).

In [src/data/dataset.py](../src/data/dataset.py), `collate_transition_batch` employs **PyG-style disjoint union batching**:
1. Node features (`faces_features`, `edges_features`) are concatenated along dimension 0 into `[Total_faces, 32]` and `[Total_edges, 16]`.
2. Graph adjacency indices are shifted by cumulative face offsets: $\text{adj}_{\text{batch}} = \text{adj}_i + \text{offset}_i$.
3. Batch index pointers (`batch_faces`, `batch_edges`) assign each node back to its batch element $b \in [0, B-1]$.
4. Multi-view images are stacked into a standard tensor $[B, 4, 3, 224, 224]$.

```python
# Test DataLoader collation with CADTransitionDataset
from torch.utils.data import DataLoader

# Load from our preprocessed dataset
dataset = CADTransitionDataset(
    root_dir=project_root / 'data/processed_data',
    split='val',
    max_samples=10,
    step_horizon=1,
)

loader = DataLoader(
    dataset,
    batch_size=2,
    shuffle=False,
    collate_fn=collate_transition_batch,
)

# Ingest first batch
batch = next(iter(loader))
curr_s = batch['current_state']
act = batch['action']
next_s = batch['next_state']

print('Collated PyTorch Batch Breakdown (Batch Size = 2):')
print('  Current State:')
print(f'    Images:                 {curr_s["images"].shape}  [B, 4, 3, 224, 224]')
print(f'    Concatenated Faces:     {curr_s["faces_features"].shape}  [Total_faces, 32]')
print(f'    Face Batch Pointers:    {curr_s["batch_faces"].shape}  [Total_faces]')
print(f'    Concatenated Edges:     {curr_s["edges_features"].shape}  [Total_edges, 16]')
print(f'    Edge Batch Pointers:    {curr_s["batch_edges"].shape}  [Total_edges]')
print(f'    Disjoint FAG Edges:     {curr_s["faces_adjacency_index"].shape}  [2, Total_E_adj]')
print('  Action Context:')
print(f'    Command IDs:            {act["cmd_id"].shape}  [B]')
print(f'    Continuous Parameters:  {act["params"].shape}  [B, 12]')
print(f'    Selected Faces Mask:    {act["selected_faces_mask"].shape}  [Total_faces]')
print(f'    Selected Edges Mask:    {act["selected_edges_mask"].shape}  [Total_edges]')
print(f'    Profile Points:         {act["profile_points_2d"].shape}  [B, 64, 2]')
print('  Next State:')
print(f'    Images:                 {next_s["images"].shape}  [B, 4, 3, 224, 224]')
print(f'    Bounding Box Delta:     {batch["bbox_delta"].shape}  [B, 6]')
print('\nBatch is 100% neural-ready for World Model & JEPA training!')
```
